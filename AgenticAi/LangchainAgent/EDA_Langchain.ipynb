```python
import os
import warnings
import pandas as pd
warnings.filterwarnings("ignore")
```

```python
#We are integrating pandas agent to perform or generate queries or code to perform EDA Tasks to work with Datasets
from langchain_experimental.agents import create_pandas_dataframe_agent
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
from dotenv import load_dotenv, find_dotenv
```

```python
#Create an Agent
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv()) #It pulls or loads Google API key from .env file
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
```

```python

#Read titanic dataset downloaded from Kaggle
titanic_df = pd.read_csv("titanic_dataset.csv")
titanic_df.head()
```

```python
#pandas dataframe agent will interact with titanic_df dataframe throuh the LLM,
agent = create_pandas_dataframe_agent(
	llm,
	titanic_df,
	verbose=True,
	agent_type = 'openai-tools',
	allow_dangerous_code=True
)
```

```python
agent.run("How Many rows are there")
```

```python
agent.run("How Many people have more than 3 siblings")
```

```python
agent.run("Which are dependent and Independent Variables")
```

```python
agent.run("Which attributes has missing values")
```
